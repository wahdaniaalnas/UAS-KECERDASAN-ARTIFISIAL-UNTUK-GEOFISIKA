## **Alur Pipeline**

```mermaid
flowchart LR
    A[DEM, hujan, geologi, tutupan lahan, inventori longsor] --> B[Preprocessing GIS]
    B --> C[Faktor pengondisi + seleksi fitur]
    C --> D1[P1 CNN-DNN]
    C --> D2[P2 TSDNN hujan time-series]
    C --> D3[P3 CTLGNet CNN+Transformer]
    C --> D4[P4 CRNN-LSTM stacking + GW-DNN]
    D1 & D2 & D3 & D4 --> E[Evaluasi: AUC, F1, Acc, RMSE]
    E --> F[Peta kerentanan + Report]
    G[P5 Review: taksonomi] -.kerangka.-> E
```


## **Skill per Paper**

| Paper | Skill yang dipelajari | Cell program |
|-------|----------------------|--------------|
| P1 Azarafza | Menyusun dataset spasial positif/negatif, normalisasi faktor, CNN–DNN, metrik AUC | `P1` |
| P2 Gao | Mengolah hujan *time-series*, menggabungkan fitur statis + temporal (dua cabang) | `P2` |
| P3 Zhao | Membuat *patch* dari raster, CNN untuk fitur lokal, self-attention untuk fitur global, kepentingan faktor (RF–Gini) | `P3` |
| P4 Guo | Stacking beberapa model sekuens, meta-learner, regresi berbobot geografis (GWR) | `P4` |
| P5 Zhang & Wang | Memetakan tugas DL (klasifikasi/deteksi/segmentasi/sekuens/hibrid), protokol evaluasi | `Evaluasi` |
